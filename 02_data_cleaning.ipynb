# Supply Chain Data Cleaning

This notebook cleans and standardizes the seven supply-chain datasets based on the issues identified during data profiling.

The raw datasets are preserved without modification. Clean copies are created for all transformations.

##3.Standardize Text Values

```python
text_cleaning_summary = []

for table_name, df in clean_tables.items():

    # Select all string columns
    string_columns = df.select_dtypes(
        include="string"
    ).columns

    for column in string_columns:
        before = df[column].copy()

        # Remove surrounding spaces and standardize uppercase
        df[column] = (
            df[column]
            .str.strip()
            .str.upper()
        )

        changed_count = (
            before.ne(df[column])
            .fillna(False)
            .sum()
        )

        text_cleaning_summary.append({
            "table": table_name,
            "column": column,
            "changed_values": changed_count
        })
```

```python
text_cleaning_summary = pd.DataFrame(text_cleaning_summary)
display(text_cleaning_summary)
```

# Data Cleaning Summary

### 1. 建立資料副本

使用 `.copy()` 建立清理用的 DataFrame，避免直接修改原始資料。

### 2. 統一欄位名稱

將不同資料表中代表相同概念的欄位統一命名，例如：

- `tpt`、`tpt_day_cnt` → `transport_days`
- `orig_port_cd` → `origin_port`
- `dest_port_cd` → `destination_port`
- `plant_id`、`wh` → `plant_code`
- `customers` → `customer`

### 3. 修正資料型態

- 將 `order_date` 轉換為日期格式。
- 將訂單、產品、工廠、港口、客戶及其他類別欄位轉為字串格式。
- 將數量、天數及產能欄位轉為整數格式。
- 將重量、費率及成本欄位轉為浮點數格式。
- 將 `order_id` 以字串格式保存，避免包含小數的識別碼被四捨五入或截斷。

### 4. 標準化文字資料

針對所有字串欄位：

- 移除文字前後的空格。
- 將代碼及類別值統一轉換為大寫。

例如：

- ` plant03 ` → `PLANT03`
- `air` → `AIR`

### 5. 處理重複資料

- 在 Freight Rates 資料表中發現 3 筆額外的完全重複資料。
- 使用 `.drop_duplicates()` 移除重複資料。
- Freight Rates 的資料筆數由 1,540 筆減少為 1,537 筆。
- 其他資料表未發現完全重複的紀錄。

### 6. 驗證數值合理性

檢查以下可能的資料品質問題：

- 負數的運輸天數
- 負數的提前或延遲出貨天數
- 非正數的訂購數量
- 負數的最低成本或運費率
- 最小重量大於最大重量
- 非正數的倉庫產能
- 非正數的倉庫處理成本

除了 2 筆零重量訂單外，其餘數值檢查皆通過。

### 7. 標記零重量訂單

- 原始資料中有 2 筆訂單的 `weight` 為 0。
- 保留這 2 筆訂單，避免影響訂單數量、客戶、產品及工廠分析。
- 新增 `weight_valid` 欄位標記重量是否有效：
  - `True`：重量大於 0，可以進行重量與運費分析。
  - `False`：重量等於 0，不納入重量與運費分析。
- 完整訂單分析使用 `orders_clean`。
- 重量與運費分析使用 `valid_weight_orders`。

### 8. 驗證出貨時間範圍

- 確認 `ship_ahead_day_count` 與 `ship_late_day_count` 沒有負數。
- 兩個欄位同時大於 0 不代表資料衝突，而是表示訂單允許的提前與延遲出貨範圍。
- 因此保留原始的 shipping-window 數值。

### 9. 驗證 Key 與資料表關聯

針對各資料表的 Primary Key、Composite Key 及表格間的關聯進行驗證：

- 檢查 `orders` 中的 `order_id` 是否唯一。
- 檢查 `plant_ports` 中的 `plant_code + port` 組合是否重複。
- 檢查 `products_per_plant` 中的 `plant_code + product_id` 組合是否重複。
- 檢查 `vmi_customers` 中的 `plant_code + customer` 組合是否重複。
- 檢查 Warehouse Capacities 與 Warehouse Costs 中的 `plant_code` 是否唯一。
- Freight Rates 沒有單一欄位可以作為唯一識別值，因此使用 carrier、路線、服務類型、運輸方式及重量區間組成 Composite Key 進行檢查。
- 確認訂單中的工廠能在產能、倉庫成本及工廠港口資料表中找到。
- 確認訂單中的 `plant_code + product_id` 組合能否對應至 Products per Plant 資料表。
- 由於 Orders 為歷史訂單，而其他資料表代表目前的供應鏈限制，因此無法對應的歷史組合將先保留並記錄，不會直接刪除。

### 10. 最終資料品質驗證

完成資料清理後，再次對七個資料表進行整體驗證：

- 確認各資料表清理後的資料筆數與欄位數。
- 再次檢查 missing values 與完全重複資料。
- 確認 `order_id` 及各資料表的 Composite Key 沒有非預期重複。
- 再次檢查負數、零值及不合理的重量區間。
- 確認工廠、產能、成本、港口及產品之間的資料關聯。
- Freight Rates 移除 3 筆完全重複資料後，由 1,540 筆減少為 1,537 筆。
- Orders 保留完整的 9,215 筆資料。
- 其中 2 筆零重量訂單以 `weight_valid = False` 標記，只在重量與運費分析時排除。
- 清理後的完整訂單主表為 `orders_clean`，重量與運費分析使用 `valid_weight_orders`。
- 完成驗證後，將清理結果匯出至 cleaned data 資料夾，供後續 EDA、SQL、成本分析及 Power BI 使用。
